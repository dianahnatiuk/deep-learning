# Розпізнавання власних цифр збереженою CNN
Завантажте в Colab файл `final_cnn_mnist.keras` та папку `my_digits/` з файлами `0.png … 9.png` (темна цифра на білому, ~квадратне фото).

```python
import numpy as np, matplotlib.pyplot as plt, glob, os
from PIL import Image
from tensorflow import keras

model = keras.models.load_model('final_cnn_mnist.keras')
# перекомпіляція (за умовою завдання)
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

def prepare(path):
    img = Image.open(path).convert('L').resize((28,28))
    arr = 255 - np.array(img)                 # інверсія кольорів
    arr = arr.astype('float32') / 255         # нормалізація
    return arr.reshape(1,28,28,1)             # додаємо вимір батчу і канал

files = sorted(glob.glob('my_digits/*.png'))
correct = 0
plt.figure(figsize=(14,3))
for k,f in enumerate(files):
    x = prepare(f)
    p = model.predict(x, verbose=0)[0]
    true = int(os.path.basename(f).split('.')[0][0])
    pred = int(p.argmax()); correct += (pred==true)
    plt.subplot(1,len(files),k+1); plt.imshow(x.squeeze(), cmap='gray'); plt.axis('off')
    plt.title(f"true {true}\npred {pred} ({p.max()*100:.0f}%)")
plt.show()
print(f"Правильно: {correct} з {len(files)}")
```
